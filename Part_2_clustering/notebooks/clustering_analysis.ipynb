```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
red_wine = pd.read_excel(
    "../data/wine data.xlsx",
    sheet_name="Red Wine"
)

red_wine.head()
```

```python
white_wine = pd.read_excel(
    "../data/wine data.xlsx",
    sheet_name="White wine"
)

white_wine.head()
```

```python
red_wine['wine_type'] = 'Red'

white_wine['wine_type'] = 'White'
```

```python
wine_df = pd.concat(
    [red_wine, white_wine],
    ignore_index=True
)

wine_df.head()
```

```python
wine_df = wine_df.loc[:, ~wine_df.columns.str.contains('^Unnamed')]
```

```python
wine_df.head()
```

```python
wine_df.shape
```

```python
X = wine_df.drop(columns=['quality', 'wine_type'])
```

```python
y = wine_df['wine_type']
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
X_scaled[:5]
```

```python
from sklearn.cluster import KMeans
```

```python
kmeans = KMeans(
    n_clusters=2,
    random_state=42
)

kmeans.fit(X_scaled)
```

```python
clusters = kmeans.labels_
```

```python
clusters[:20]
```

```python
y_numeric = y.map({
    'Red': 0,
    'White': 1
})
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score
```

```python
pd.concat()
```

```python
cm = confusion_matrix(y_numeric, clusters)

cm
```

```python
accuracy = accuracy_score(y_numeric, clusters)

print("Accuracy:", accuracy)
```

```python
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.xlabel("Predicted Cluster")
plt.ylabel("Actual Wine Type")
plt.title("Confusion Matrix")

plt.savefig("../outputs/confusion_matrix.png")


plt.show()
```

```python
wcss = []
```

```python
for k in range(1, 11):
    
    kmeans = KMeans(
        n_clusters=k,
        random_state=42
    )
    
    kmeans.fit(X_scaled)
    
    wcss.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    range(1,11),
    wcss,
    marker='o'
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.title("Elbow Method")

plt.savefig("../outputs/elbow_method.png")

plt.show()
```

```python
from sklearn.metrics import silhouette_score
```

```python
silhouette_scores = []

for k in range(2, 11):
    
    kmeans = KMeans(
        n_clusters=k,
        random_state=42
    )
    
    cluster_labels = kmeans.fit_predict(X_scaled)
    
    score = silhouette_score(
        X_scaled,
        cluster_labels
    )
    
    silhouette_scores.append(score)
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    range(2,11),
    silhouette_scores,
    marker='o'
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Analysis")

plt.savefig("../outputs/silhouette_analysis.png")


plt.show()
```

```python
from scipy.cluster.hierarchy import linkage, dendrogram
```

```python
single_linkage = linkage(
    sample_data,
    method='single'
)
```

```python
plt.figure(figsize=(12,6))

dendrogram(single_linkage)

plt.title("Single Linkage Dendrogram")
plt.xlabel("Observations")
plt.ylabel("Distance")

plt.savefig("../outputs/single_linkage.png")

plt.show()
```

```python
complete_linkage = linkage(
    sample_data,
    method='complete'
)

plt.figure(figsize=(12,6))

dendrogram(complete_linkage)

plt.title("Complete Linkage Dendrogram")
plt.xlabel("Observations")
plt.ylabel("Distance")

plt.savefig("../outputs/complete.png")

plt.show()
```

```python
average_linkage = linkage(
    sample_data,
    method='average'
)

plt.figure(figsize=(12,6))

dendrogram(average_linkage)

plt.title("Average Linkage Dendrogram")
plt.xlabel("Observations")
plt.ylabel("Distance")

plt.savefig("../outputs/average.png")

plt.show()
```
